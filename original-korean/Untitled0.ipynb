```python
df_total = pd.read_csv('/content/drive/MyDrive/Econometrics_Project_PNU/nba20232024_total.txt', delimiter = ',')
df_advanced = pd.read_csv('/content/drive/MyDrive/Econometrics_Project_PNU/nba20232024_advanced.txt', delimiter = ',')
```

```python
merged_df = pd.merge(df_total, df_advanced, on = "Player")

columns_need = ['Player', 'G_x', 'PER', 'PTS', 'AST', 'TRB', 'STL', 'BLK']
merged_df_clean = merged_df[columns_need]

merged_df_clean['PTS_per_game'] = merged_df_clean['PTS'] / merged_df_clean['G_x']
merged_df_clean['AST_per_game'] = merged_df_clean['AST'] / merged_df_clean['G_x']
merged_df_clean['TRB_per_game'] = merged_df_clean['TRB'] / merged_df_clean['G_x']
merged_df_clean['STL_per_game'] = merged_df_clean['STL'] / merged_df_clean['G_x']
merged_df_clean['BLK_per_game'] = merged_df_clean['BLK'] / merged_df_clean['G_x']

per_stats = ['PTS_per_game', 'AST_per_game', 'TRB_per_game', 'STL_per_game', 'BLK_per_game']
df_stats = merged_df_clean[['Player','PER'] + per_stats]
```

```python
columns_need_stats = ['PER', 'PTS_per_game', 'AST_per_game', 'TRB_per_game', 'STL_per_game', 'BLK_per_game']
df_stats_clean = df_stats[columns_need_stats]
```

```python
df_stats_clean = df_stats_clean.iloc[:100]
```
